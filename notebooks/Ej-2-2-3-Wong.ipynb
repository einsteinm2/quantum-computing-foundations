### **Ejemplo**:

*Crear y Verificar el estado inicial de un solo qubit en Qiskit*

<br/>

Por defecto, cuando creamos un circuito en Qiskit (simulador), cada qubit se inicializa en el estado base $\ket{0} = \vzero$. En un computador cuántico físico esto no ocurre de manera ideal, los qubits no siempre comienzan exactamente en $\ket{0}$ debido a *ruido térmico*, *fluctuaciones electromagnéticas* y otros factores.

```python
# -------------------------------------------------------------
# Ejemplo: Verificar el estado inicial de un qubit en Qiskit
# -------------------------------------------------------------

# Importamos las librerías necesarias
from qiskit import QuantumCircuit #Para crear un circuito cuántico
from qiskit.quantum_info import Statevector #Para evaluar estado del circuito
from qiskit.visualization import plot_bloch_multivector #Para visualizar geometricamente un qubit

# Creamos un circuito con 1 qubit
qc = QuantumCircuit(1)

# Obtenemos el vector de estado del circuito
state = Statevector.from_instruction(qc)

# Mostramos el vector numérico
print("Vector de estado inicial: ", state.data,"\n")

# Visualizamos el estado en la esfera de Bloch
plot_bloch_multivector(state, figsize=(3, 3))
```
