## Ejercicio 2.10 - Wong

Un qubit se encuentra en el estado $\frac{1}{2}\ket{0}-\frac{\sqrt{3}}{2}\ket{1}$

1.   Si el estado es medido en la base $Z$, que estados puedes obtener y con que probabilidad?
2.   Expresa el estado del qubit en terminos de $\ket{+}$ y $\ket{-}$
3.   Si el estado es medido en la base $\{ \ket{+},\ket{-} \}$, que estados puedes obtener y con que probabilidad?

**Sol.**

1. En la base $Z$ podemos obtener el estado $\ket{0}$ con una probabilidad de $\left|\frac{1}{2}\right|^2=\frac{1}{4}\approx 0.25 \:$ o el estado $\ket{1}$ con una probabilidad de $\left|-\frac{\sqrt{3}}{2}\right|^2=\frac{3}{4} \approx 0.75$
2. Si aplicamos la compuerta $H$ a $\frac{1}{2}\ket{0}-\frac{\sqrt{3}}{2}\ket{1}$ tenemos la siguiente igualdad


\begin{align*}
H\left(\frac{1}{2}\ket{0}-\frac{\sqrt{3}}{2}\ket{1}\right)&=\frac{1}{2}H\ket{0}-\frac{\sqrt{3}}{2}H\ket{1}=\frac{1}{2}\had{+}-\frac{\sqrt{3}}{2}\had{-}\\
&=\left( \frac{1-\sqrt{3}}{2\sqrt{2}} \right)\ket{0} \: +  \left( \frac{1+\sqrt{3}}{2\sqrt{2}} \right)\ket{1}
\end{align*}


Ahora si aplicamos $H$ nuevamente a la igualdad tenemos que

\begin{align*}
H^2\left(\frac{1}{2}\ket{0}-\frac{\sqrt{3}}{2}\ket{1}\right)&=\left( \frac{1-\sqrt{3}}{2\sqrt{2}} \right)H\ket{0} \: +  \left( \frac{1+\sqrt{3}}{2\sqrt{2}} \right)H\ket{1}\\
I\left(\frac{1}{2}\ket{0}-\frac{\sqrt{3}}{2}\ket{1}\right)&=\left( \frac{1-\sqrt{3}}{2\sqrt{2}} \right)\ket{+} \: +  \left( \frac{1+\sqrt{3}}{2\sqrt{2}} \right)\ket{-}\\
\frac{1}{2}\ket{0}-\frac{\sqrt{3}}{2}\ket{1}&=\left( \frac{1-\sqrt{3}}{2\sqrt{2}} \right)\ket{+} \: +  \left( \frac{1+\sqrt{3}}{2\sqrt{2}} \right)\ket{-}\\
\end{align*}

<br/>

3. En la base $\{ \ket{+},\ket{-} \}$ podemos obtener el estado $\ket{0}$ con una probabilidad de $\left|\frac{1-\sqrt{3}}{2\sqrt{2}}\right|^2=\frac{2-\sqrt{3}}{4}\approx 0.07 \:$ o el estado $\ket{1}$ con una probabilidad de $\left|\frac{1+\sqrt{3}}{2\sqrt{2}}\right|^2=\frac{2+\sqrt{3}}{4}\approx 0.93$


## 3. Verificación con Qiskit
#### Caso 1.

```python
# -------------------------------------------------------------
# Ejercicio: 2.10 (1.) Wong
# -------------------------------------------------------------

from qiskit import QuantumCircuit, transpile
from qiskit.quantum_info import Statevector
from qiskit_aer import AerSimulator
from qiskit.visualization import plot_histogram
import numpy as np
import matplotlib.pyplot as plt

#Creamos un circuito con un solo qubit y 1 bit clásico para medir
qc = QuantumCircuit(1, 1)

#Inicializamos el estado del qubit a
alpha = 1/2
beta = -1*np.sqrt(3)/2
qc.initialize([alpha, beta], 0)

#Evaluamos el estado del qubit
state = Statevector.from_instruction(qc)

#Añadimos la medición del qubit al bit clásico
qc.measure(0,0)

# Simulador
sim = AerSimulator()

# Transpilamos y ejecutamos el circuito varias veces
tqc = transpile(qc, sim)
result = sim.run(tqc, shots=2000).result()

# Mostramos el vector numérico
print("\nVector de estado inicial: ", state.data)

# Conteos de resultados
counts = result.get_counts()
print("\nResultados de la medición:", counts)

# Histograma de resultados
plot_histogram(counts, figsize=(4, 3.5))
```

#### Caso 2.
En este caso, para verificarlo en Qiskit lo que haremos es aplicar una vez l acompuerta $H$ a $\frac{1}{2}\ket{0}-\frac{\sqrt{3}}{2}\ket{1}$, así podremos conocer sus amplitudes en la base $\{ \ket{+},\ket{-} \}$, note que para llevarlo a la base $\{ \ket{+},\ket{-} \}$, tendríamos que aplicar nuevamente la compuerta $H$ tal y como se mostró arriba, no lo aplicamos porque qiskit los estados los da en la base $Z$ por lo que , con una vez que apliquemos la compuerta H es sufiente para conocer las amplitudes en la base $\{ \ket{+},\ket{-} \}$

Para expresar el estado
$\frac{1}{2}\ket{0}-\frac{\sqrt{3}}{2}\ket{1}$
en la base $\{\ket{+},\ket{-}\}$, obtendremos solo sus amplitudos. Para esto basta con aplicar una vez la compuerta $H$.

Qiskit siempre expresa los estados en la base computacional $\{\ket{0}, \ket{1}\}$, por lo que ese único Hadamard realiza la transformación necesaria y el resultado corresponde directamente a las amplitudes en la base $\{\ket{+},\ket{-}\}$. (**Mirar el desarrollo algebraico arriba**)

```python
# -------------------------------------------------------------
# Ejercicio: 2.10 (2.) Wong
# -------------------------------------------------------------

from qiskit import QuantumCircuit
from qiskit.quantum_info import Statevector

# Creamos un circuito cualquiera
qc = QuantumCircuit(1)

#Inicializamos el estado del qubit a
alpha = 1/2
beta = -1*np.sqrt(3)/2
qc.initialize([alpha, beta], 0)

#Aplicamos la compuerta H
qc.h(0)


#Evaluamos el estado del qubit
state = Statevector.from_instruction(qc)
print("Estado:", state.data)
```

#### Caso 3.

```python
# -------------------------------------------------------------
# Ejercicio: 2.10 (3.) Wong
# -------------------------------------------------------------

from qiskit import QuantumCircuit, transpile
from qiskit.quantum_info import Statevector
from qiskit_aer import AerSimulator
from qiskit.visualization import plot_histogram
import numpy as np
import matplotlib.pyplot as plt

#Creamos un circuito con un solo qubit y 1 bit clásico para medir
qc = QuantumCircuit(1, 1)

#Inicializamos el estado del qubit a
alpha = 1/2
beta = -1*np.sqrt(3)/2
qc.initialize([alpha, beta], 0)

#Aplicamos la compuerta H
qc.h(0)

#Evaluamos el estado del qubit
state = Statevector.from_instruction(qc)

#Añadimos la medición del qubit al bit clásico
qc.measure(0,0)

# Simulador
sim = AerSimulator()

# Transpilamos y ejecutamos el circuito varias veces
tqc = transpile(qc, sim)
result = sim.run(tqc, shots=2000).result()

# Mostramos el vector numérico
print("\nVector de estado inicial: ", state.data)

# Conteos de resultados
counts = result.get_counts()
print("\nResultados de la medición:", counts)

# Histograma de resultados
plot_histogram(counts, figsize=(4, 3.5))
```

Aunque fue medido en la base $Z$ las amplitudes son las que corresponden al estado en la base $\{\ket{+}, \ket{-}\}$.
