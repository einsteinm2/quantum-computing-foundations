## Ejercicio 2.33 - Wong

Pruebe que $HZH=X$ mostrando que $HZH\ket{0}$ y $X\ket{0}$ resultan en el mismo estado y $HZH\ket{1}$ y $X\ket{1}$ también.

**Sol.**
\begin{align*}
HZH\ket{0} &= HZ\left(\had{+}\right) = \frac{1}{\sqrt{2}}H\left(Z\ket{0}+Z\ket{1}\right)\\
          &= \frac{1}{\sqrt{2}}H\left(\ket{0}-\ket{1}\right)
          = H\ket{-} = \ket{1}
\end{align*}
\begin{align*}
HZH\ket{1} &= HZ\left(\had{-}\right) = \frac{1}{\sqrt{2}}H\left(Z\ket{0}-Z\ket{1}\right)\\
          &= \frac{1}{\sqrt{2}}H\left(\ket{0}+\ket{1}\right)
          = H\ket{+} = \ket{0}
\end{align*}

En ambos casos coincide con la acción de la compuerta $X$, por lo que $HZH=X$.


## 3. Verificación con Qiskit


```python
# -------------------------------------------------------------
# Ejercicio: 2.33 Wong
# -------------------------------------------------------------

from qiskit import QuantumCircuit
from qiskit.quantum_info import Statevector

#Creamos un circuito con un solo qubit
qc = QuantumCircuit(1)

#Añadimos una compuerta H, Z, H
qc.h(0)
qc.z(0)
qc.h(0)

#Evaluamos el estado del qubit
state0 = Statevector.from_instruction(qc)
print('HZH|0> =', state0.data)
```

Note que el resultado `[0.+0.j 1.+0.j]` es igual a $X\ket{0}=\ket{1}$. Ahora veamos el caso para $HZH\ket{1}$.


```python
# -------------------------------------------------------------
# Ejercicio: 2.33 Wong
# -------------------------------------------------------------

from qiskit import QuantumCircuit
from qiskit.quantum_info import Statevector

#Creamos un circuito con un solo qubit
qc = QuantumCircuit(1)

#Inicializamos el estado del qubit a |1>
alpha = 0
beta = 1
qc.initialize([alpha, beta], 0)

#Añadimos una compuerta H, Z, H
qc.h(0)
qc.z(0)
qc.h(0)

#Evaluamos el estado del qubit
state1 = Statevector.from_instruction(qc)
print('HZH|1> =', state1.data)
```

Note que el resultado `[1.+0.j 0.+0.j]` es igual a $X\ket{1}=\ket{0}$.


## 4. Visualización


```python
# Visualizamos estados en la esfera de Bloch
from qiskit.visualization import plot_bloch_multivector
plot_bloch_multivector(state0, title='HZH|0⟩', figsize=(3, 3))
```

```python
plot_bloch_multivector(state1, title='HZH|1⟩', figsize=(3, 3))
```

## 5. Conclusión

La igualdad $HZH=X$ es una **identidad de circuitos**, útil para simplificar y optimizar circuitos cuánticos sin cambiar su efecto.
